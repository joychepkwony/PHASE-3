```python
#handling missing values

threshold = 0.5  

# Calculate the percentage of missing values per column
missing_percent = merged_data.isnull().mean()

# Drop columns where missing percentage exceeds the threshold
merged_data = merged_data.drop(columns=missing_percent[missing_percent > threshold].index)

# Verify dropped columns
print(merged_data.isnull().sum())





```

```python
#HANDLING OUTLIERS 
# Identify numerical columns
numerical_cols = merged_data.select_dtypes(include=["int64", "float64"]).columns

# Boxplot visualization to detect outliers
plt.figure(figsize=(15, 8))
outlier_cols = ["amount_tsh", "gps_height", "longitude", "num_private", "population", "construction_year"]

for i, col in enumerate(outlier_cols, 1):
    plt.subplot(2, 3, i)
    sns.boxplot(x=merged_data[col])
    plt.title(f"Boxplot of {col}")

plt.tight_layout()
plt.show()
```

```python
# Handling outliers using the IQR method
def remove_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    return df
```

```python
# Applying outlier removal to selected columns
for col in ["amount_tsh", "num_private", "population"]:
    merged_data = remove_outliers(merged_data, col)
```

```python
# Fixing  incorrect values

merged_data.loc[merged_data["gps_height"] < 0, "gps_height"] = merged_data["gps_height"].median()

# Replacing longitude = 0 with the median of non-zero values
median_longitude = merged_data.loc[merged_data["longitude"] != 0, "longitude"].median()
merged_data.loc[merged_data["longitude"] == 0, "longitude"] = median_longitude

# Replacing construction_year = 0 with median of non-zero years
median_year = merged_data[merged_data["construction_year"] > 0]["construction_year"].median()
merged_data["construction_year"] = merged_data["construction_year"].replace(0, median_year)


print(merged_data.describe())
```
